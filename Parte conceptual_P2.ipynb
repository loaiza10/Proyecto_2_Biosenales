# **Clasificación de datos BCI**

### **Indices del EEG que mejoran la clasificación de estados en BCI.**

La densidad espectral de potencia (PSD) es una de las técnicas más utilizadas en sistemas Brain-Computer Interface (BCI) para analizar la actividad cerebral asociada a tareas de imaginación motora. Sin embargo, debido a la alta complejidad y variabilidad de las señales EEG, el uso exclusivo de características espectrales puede resultar insuficiente para lograr una clasificación robusta entre diferentes estados cerebrales. Por esta razón, en la literatura científica se han propuesto índices adicionales que complementan la información obtenida mediante PSD, incorporando características espaciales, temporales y no lineales del EEG, con el fin de mejorar el desempeño de los sistemas BCI [1]–[3].

En el presente proyecto, el interés principal se centra en diferenciar estados de imaginación motora de mano izquierda y derecha respecto al estado de reposo, utilizando señales EEG registradas en las regiones sensoriomotoras C3, C4 y Cz. Debido a esto, se seleccionaron tres índices ampliamente utilizados en aplicaciones BCI y compatibles con implementación en Python: Common Spatial Patterns (CSP), entropía espectral y parámetros de Hjorth.

Uno de los métodos más utilizados en sistemas BCI es **Common Spatial Patterns (CSP)**. Este método corresponde a una técnica de filtrado espacial que maximiza las diferencias entre dos clases de señales EEG, encontrando combinaciones espaciales de canales que aumentan la varianza de una condición y reducen la de otra. En tareas de imaginación motora, CSP permite resaltar la activación cortical asociada al movimiento imaginado de la mano izquierda o derecha, especialmente en regiones motoras contralaterales. Las características obtenidas mediante CSP suelen calcularse a partir del logaritmo de la varianza de las señales transformadas espacialmente. Este método es ampliamente utilizado debido a su alta capacidad discriminativa en sistemas BCI basados en imaginación motora, mejorando significativamente la clasificación entre diferentes tareas motoras imaginadas [1].

Ramoser et al. demostraron que CSP mejora significativamente la clasificación de movimientos imaginados utilizando señales EEG multicanal, convirtiéndose en una de las técnicas más utilizadas en sistemas BCI modernos [1]. En Python, este método puede implementarse mediante la librería `MNE`, utilizando el módulo `mne.decoding.CSP` [4].

Otro índice relevante es la **entropía espectral**, una medida basada en teoría de la información que cuantifica el grado de complejidad o desorden de la señal EEG en el dominio de la frecuencia. Esta medida se calcula a partir de la PSD normalizada y permite evaluar cómo se distribuye la energía de la señal entre las diferentes frecuencias. Valores bajos de entropía indican una señal más organizada y dominada por ciertos ritmos cerebrales, mientras que valores altos representan una distribución espectral más compleja. Durante tareas de imaginación motora se producen cambios en los ritmos sensoriomotores μ y β, modificando la organización espectral respecto al estado de reposo. Por ello, la entropía espectral puede utilizarse para mejorar la diferenciación entre imaginación motora y reposo, aportando información no lineal que no puede observarse únicamente mediante potencia espectral [2].

Diversos estudios han mostrado que las características basadas en entropía mejoran el desempeño de los clasificadores en sistemas BCI debido a su capacidad para representar cambios complejos en la dinámica cerebral [2]. En Python, esta medida puede implementarse utilizando librerías como `antropy` o funciones basadas en `scipy.signal.welch`. [5]

Adicionalmente, se seleccionaron los **parámetros de Hjorth**, ampliamente utilizados en análisis EEG por su simplicidad computacional y capacidad para describir la dinámica temporal de las señales. Estos parámetros incluyen actividad, movilidad y complejidad. La actividad corresponde a la varianza de la señal; la movilidad representa el contenido frecuencial promedio; y la complejidad describe la variación de frecuencia de la señal a lo largo del tiempo. En sistemas BCI, los parámetros de Hjorth permiten analizar cambios dinámicos asociados a imaginación motora respecto al reposo y pueden calcularse fácilmente sobre canales motores como C3, C4 y Cz [3].

Los parámetros de Hjorth son especialmente útiles en aplicaciones en tiempo real debido a su bajo costo computacional y su capacidad para representar características temporales relevantes del EEG [3]. En Python, estos parámetros pueden calcularse mediante librerías como `antropy` o implementaciones directas utilizando `NumPy`. [5]

En conjunto, estos índices complementan el análisis realizado previamente mediante densidad espectral de potencia, ERD/ERS y coherencia. Mientras la PSD analiza la distribución de energía en determinadas bandas de frecuencia, técnicas como CSP incorporan información espacial, la entropía espectral aporta medidas de complejidad y los parámetros de Hjorth describen la dinámica temporal de la señal. La combinación de estas características puede mejorar significativamente la clasificación entre imaginación motora y reposo en sistemas BCI no invasivos.
